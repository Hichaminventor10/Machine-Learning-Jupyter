# Régression Linéaire 

```python
import numpy as np
import numpy.random as rnd
from matplotlib import pyplot as plt

np.random.seed(42)
```

### Construction du jeu de donnée

```python
m = 100
x = 6 * np.random.rand(m) - 3
x.sort()
x=x.reshape(m,1)
y =2 + x +  x**2 + np.random.randn(m, 1)
```

### 1 Afficher le nuage des données, (limiter x entre -3 et 3 et y entre 0 et 10)

```python

plt.scatter(x, y)


plt.xlim(-3, 3)
plt.ylim(0, 10)


plt.xlabel("x")
plt.ylabel("y")
plt.title("Nuage de données")

plt.show()
```

### 2 Construire la matrice $X$ en utilisant $x$ et $x^2$ (utiliser `np.hstack()`)

```python
X = np.hstack((x, x**2)) #plusieurs arrays horizontalement 

print(X)
print(X.shape)
```

### 3 Créer une régression linéaire et l'ajuster aux données puis afficher les paramètres de votre modèle

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()


model.fit(X, y)


print("Intercept (biais) :", model.intercept_)
print("Coefficients :", model.coef_)
```

### 3 Afficher dans un meme graphique le nuage des points et votre modèle

```python

y_pred = model.predict(X)


plt.scatter(x, y, label="Données réelles")


plt.plot(x, y_pred, color='red', label="Modèle de régression")


plt.xlim(-3, 3)
plt.ylim(0, 10)




plt.xlabel("x")
plt.ylabel("y")
plt.title("Régression polynomiale")
plt.legend()
plt.show()
```

### 6 Afficher à nouveau le nuage des points et le modèle

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(x, y, color="steelblue", label="Données réelles")
plt.plot(x, y_pred, color="red", linewidth=2, label="Modèle ajusté")
plt.xlim(-3, 3)
plt.ylim(0, 10)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Nuage de points et modèle de régression")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### 7 Créer une matrice $X_{poly}$ en utilisant `PolynomialFeatures` avec `degree=2`

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(x)

print(X_poly)
print(X_poly.shape)
```

### 8 Ajuster une régression linéaire aux données puis afficher ses paramètres 

```python
from sklearn.linear_model import LinearRegression

model_poly = LinearRegression()
model_poly.fit(X_poly, y)

print("Intercept :", model_poly.intercept_)
print("Coefficients :", model_poly.coef_)
```

### 9 Afficher le nuage de points et le modèle

```python
y_pred_poly = model_poly.predict(X_poly)

plt.scatter(x, y, label="Données réelles")
plt.plot(x, y_pred_poly, color="red", label="Modèle polynomial")

plt.xlim(-3, 3)
plt.ylim(0, 10)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Régression polynomiale de degré 2")
plt.legend()
plt.show()
```

### 10 Importer `StandardScaler` à partir de `sklearn.preprocessing`, créer un jeu de données en appliquant `PolynomialFeatures` de degré `300` à $x$ puis appliquer `StandardScaler` aux données résultantes et engregistrer le résultat dans la variable $X_{300}$

```python
from sklearn.preprocessing import PolynomialFeatures, StandardScaler

poly300 = PolynomialFeatures(degree=300)
X300 = poly300.fit_transform(x)

scaler = StandardScaler()
X300 = scaler.fit_transform(X300)

print(X300)
print(X300.shape)
```

### 11 Ajuster une régression linéaire à $X_{300}$ puis afficher le modèle et le nuage de points

```python
model300 = LinearRegression() # modele 
model300.fit(X300, y) # jusqua degre 300 

y_pred300 = model300.predict(X300)

plt.scatter(x, y, label="Données réelles") #affiche nuage 
plt.plot(x, y_pred300, color="red", label="Modèle polynomial degré 300")

plt.xlim(-3, 3)
plt.ylim(0, 10)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Régression polynomiale de degré 300")
plt.legend()
plt.show()
```

### 12 Interpréter 

```python
from sklearn.metrics import mean_squared_error, r2_score

mse_train300 = mean_squared_error(y, y_pred300)
r2_train300 = r2_score(y, y_pred300)

print("MSE du modèle de degré 300 :", mse_train300)
print("R² du modèle de degré 300 :", r2_train300)
print("\nInterprétation : un degré très élevé donne un modèle très flexible qui peut suivre le bruit des données d'apprentissage. Sans jeu de test, on ne peut pas mesurer correctement sa généralisation.")
```

# Régularisation

### 13 Importer  `Ridge` puis ajuster le modèle aux données $X_{300}$

```python
from sklearn.linear_model import Ridge

ridge300 = Ridge(alpha=1.0)
ridge300.fit(X300, y)

y_pred_ridge300 = ridge300.predict(X300)
```

### 14 Afficher le nuage de points ainsi que les deux derniers modèles ajustés à $X_{300}$ 

```python
plt.figure(figsize=(10, 6))
plt.scatter(x, y, label="Données réelles")
plt.plot(x, y_pred300, color="red", label="Régression polynomiale degré 300")
plt.plot(x, y_pred_ridge300, color="green", linewidth=2, label="Ridge degré 300, alpha=1")
plt.xlim(-3, 3)
plt.ylim(0, 10)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Régression polynomiale degré 300 avec et sans régularisation")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### 15 Utiliser différents modèle `Ridge` avec les valeurs suivantes $\{ 0.0001, 0.01, 1, 1000 \}$ pour le paramètre `alpha` 

```python
alphas = [0.0001, 0.01, 1, 1000]
ridge_models = {}
ridge_predictions = {}

for alpha in alphas:
    ridge = Ridge(alpha=alpha)
    ridge.fit(X300, y)
    ridge_models[alpha] = ridge
    ridge_predictions[alpha] = ridge.predict(X300)
```

### 16 Afficher dans un meme graphique le nuages des points et les modèles obtenus. Déduire le role du paramètre alpha

```python
plt.figure(figsize=(12, 7))
plt.scatter(x, y, color="black", s=20, label="Données réelles")

for alpha, predictions in ridge_predictions.items():
    plt.plot(x, predictions, linewidth=2, label=f"Ridge (alpha={alpha})")

plt.xlim(-3, 3)
plt.ylim(0, 10)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Influence du paramètre alpha sur la régression Ridge")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("Un alpha élevé pénalise davantage les coefficients et produit un modèle plus régularisé.")
```

### 17 Utiliser le modèle `SGDRegressor` avec `penalty="l2"`

```python
from sklearn.linear_model import SGDRegressor

sgd_model = SGDRegressor(
    penalty="l2",
    max_iter=5000,
    tol=1e-4,
    random_state=42
)
sgd_model.fit(X300, y.ravel())
y_pred_sgd = sgd_model.predict(X300)

print("MSE du modèle SGDRegressor :", mean_squared_error(y, y_pred_sgd))
print("R² du modèle SGDRegressor :", r2_score(y, y_pred_sgd))
```

### 18 Quels sont les paramètres importants `SGDRegressor`

```python
print("Paramètres importants de SGDRegressor :")
print("- penalty : type de régularisation ; ici, 'l2'.")
print("- alpha : force de la régularisation.")
print("- max_iter : nombre maximal de passages sur les données.")
print("- tol : seuil d'arrêt lorsque l'amélioration devient faible.")
print("- learning_rate : stratégie de mise à jour du taux d'apprentissage.")
print("- eta0 : taux d'apprentissage initial lorsque la stratégie l'utilise.")
print("- random_state : garantit des résultats reproductibles lorsque l'aléatoire intervient.")
```
